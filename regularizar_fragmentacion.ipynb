# Regularizar la fragmentación

El original (`RIconduitex5_5.py`) **no se toca**. Acá se ve la rampa `s_eps` y cómo mezcla las fuerzas. El solver con la rampa adentro es `RIconduitex5_5_reg.py` (mismo tiro; hay que correrlo donde ya te funciona IDA).


```python
from fragmentacion_reg import s_frag, mezclar, EPS_FRAG
import numpy as np
import matplotlib.pyplot as plt

print('eps por defecto', EPS_FRAG)
```

La rampa. El switch original es el escalón negro; las otras curvas son `s_eps` para distintos anchos.


```python
phicrit = 0.70
phi = np.linspace(0.40, 0.95, 400)
switch = (phi >= phicrit).astype(float)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(phi, switch, 'k', linewidth=2, label='switch (original)')
for eps, ls in [(0.01, '--'), (0.04, '-'), (0.10, ':')]:
    ax.plot(phi, s_frag(phi, phicrit, eps), ls, linewidth=2, label=f'eps = {eps}')
ax.axvline(phicrit, color='gray', linewidth=1)
ax.set_xlabel('phi', fontweight='bold')
ax.set_ylabel('s_eps', fontweight='bold')
ax.legend()
ax.set_title('peso de la rama fragmentada')
plt.tight_layout()
plt.show()
```

Fuerzas mezcladas. Números de juguete, para ver el salto vs la rampa. `Fmw` se apaga, `Fgw` se prende.


```python
Fmw_v, Fgw_v, Fmg_v = 80.0, 0.0, 15.0
Fmw_f, Fgw_f, Fmg_f = 0.0, 25.0, 40.0
s = s_frag(phi, phicrit, 0.04)

fig, ax = plt.subplots(1, 3, figsize=(12, 4))
ax[0].plot(phi, mezclar(switch, Fmw_v, Fmw_f), 'k', linewidth=2, label='switch')
ax[0].plot(phi, mezclar(s, Fmw_v, Fmw_f), linewidth=2, label='eps=0.04')
ax[0].set_title('Fmw (pared-fundido)')
ax[1].plot(phi, mezclar(switch, Fgw_v, Fgw_f), 'k', linewidth=2)
ax[1].plot(phi, mezclar(s, Fgw_v, Fgw_f), linewidth=2)
ax[1].set_title('Fgw (pared-gas)')
ax[2].plot(phi, mezclar(switch, Fmg_v, Fmg_f), 'k', linewidth=2)
ax[2].plot(phi, mezclar(s, Fmg_v, Fmg_f), linewidth=2)
ax[2].set_title('Fmg (arrastre)')
for a in ax:
    a.axvline(phicrit, color='gray', linewidth=1)
    a.set_xlabel('phi', fontweight='bold')
ax[0].legend()
plt.tight_layout()
plt.show()
```

Cuando `eps -> 0` la rampa vuelve al switch. Eso es el chequeo de que no cambiamos la física, solo el salto.


```python
s_chico = s_frag(phi, phicrit, 1e-4)
print('max |s_eps - switch| con eps=1e-4:', float(np.max(np.abs(s_chico - switch))))
print('max |s_eps - switch| con eps=0.04:', float(np.max(np.abs(s - switch))))
```
